```python
import numpy as np

arr = np.array([10,20,30,40,50])
result = arr > 25
print(result)
```

```python
print(arr < 30)
```

```python
print(arr == 30)
```

```python
print(arr != 30)
```

```python
arr = np.array([10,20,30,40,50])
result = arr[arr > 25]
print(result)
```

```python
arr = np.array([10,20,30,40,50])
result = arr[(arr >= 20) & (arr <= 40)]
print(result)
```

```python
result = arr[(arr < 20) | (arr > 40)]

print(result)
```

```python
marks = np.array([71,72,73,84,85,86,87])

passed_with_high_marks = marks[marks >= 80]
print("All marks: ",marks)
print("marks >= 80: ",passed_with_high_marks)
```

```python
selected = marks[(marks >= 70) & (marks <= 90)]

print(selected)
```

```python
matrix = np.array([
    [10,60,90],
    [45,90,30]
])
print(matrix > 50)
```

```python
print(matrix[matrix > 50])
```
