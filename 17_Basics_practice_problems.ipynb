```python
import numpy as np
arr = np.arange(1,7)
print("original: ",arr)
new_arr = arr.reshape(2, 3)
print("reshaped: ",new_arr)
print("shape: ",new_arr.shape)
```

```python
arr = np.arange(1, 13)

print(arr.reshape(3, 4))
print()
print(arr.reshape(4, 3))
print()
print(arr.reshape(2, 2, 3))
print()
```

```python
arr = np.arange(1, 13)

print(arr.reshape(3, -1))
print()
print(arr.reshape(-1, 4))
```

```python
arr = np.arange(1, 7)
arr2 = arr.reshape(2, 3)
print(arr2.shape)
print(arr2.ndim)
print(arr2.size)
```

```python
arr = np.arange(1, 25)

arr3d = arr.reshape(2, 3, 4)

print(arr3d)

print(arr3d.shape)
```
