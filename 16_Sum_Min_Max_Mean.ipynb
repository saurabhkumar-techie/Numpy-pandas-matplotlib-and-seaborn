```python
import numpy as np
```

```python
arr = np.array([10,20,30,40,50])

print(np.sum(arr))
```

```python
arr = np.array([12,45,7,89,23])
print("minimum: ",np.min(arr))
print("maximum: ",np.max(arr))
```

```python
marks = np.array([70,80,90,60,100])

print("average: ",np.mean(marks))
```

```python
#middle value
arr = np.array([10,20,30,50,40])
print(np.median(arr)) #median means sorted data ki middle value
print(np.median([10,20,30,40]))
```

```python
#standard Deviation
arr = np.array([10,20,30,40,50])
print(np.std(arr))
```

```python
#variance
arr = np.array([10,20,30,40,50])
print("variance: ",np.var(arr))
print("standard deviation: ",np.std(arr))
```

```python
marks = ([
    [80,70,90],
    [60,85,75],
    [95,90,88]
])
print("total: ",np.sum(marks))
print("column totals: ",np.sum(marks, axis=0))
print("rows total: ",np.sum(marks,axis=1))
```

```python
arr = np.array([10,20,30,40,50])
print("product: ",np.prod(arr))
print("range: ",np.ptp(arr))
print("cumulative sum: ",np.cumsum(arr))
print("cumulative product: ",np.cumprod(arr))
```
